# Lesson 01｜Python 基礎輸入與輸出

**學習主題：** `print()`、`input()`、字串與數字、`sep`、`end`  
**學習目標：** 能輸出文字、接收使用者輸入，並控制輸出格式。

## 1. `print()`：將內容輸出到畫面

`print()` 用來顯示資料；括號內可以放文字、數字等資料。沒有提供內容時，會輸出一個空行。

````python
print()  # 輸出空行
print('hi')  # 輸出文字 hi
````

### 數字與字串

- `123` 是整數（`int`）。
- `'123'` 是字串（`str`）。
- 字串通常使用單引號或雙引號包住。

````python
print(123)
print('123')
print('hi')
print('你好')
````

## 2. `input()`：取得使用者輸入

`input()` 會等待使用者輸入並按 Enter，**回傳輸入的字串**。可以在括號內放提示文字。

````python
input()  # 執行後請輸入文字，再按 Enter
````

````python
input('請輸入文字：')  # 提示文字會顯示在輸入位置
````

### 將輸入直接印出

`print(input())` 先取得輸入，再將結果輸出。

````python
print(input())
````

````python
print('hello,', input())
````

## 3. `print()` 的輸出格式

### `end`：控制結尾

`print()` 預設以換行結尾（`end='\n'`）；改成 `end=' '` 可以讓下一次輸出接在同一行，並用空格分隔。

````python
print(122, end=' ')
print(123)
print(124)
# 結果：第一行 122 123；第二行 124
````

### `sep`：控制多個參數之間的分隔符號

預設 `sep=' '`（空格）；設定 `sep='\n'` 可讓每個參數分行輸出。

````python
print(122, 123, 124, sep='\n')
````

## 4. 課堂練習

請先自行嘗試，再參考各題下方的示範解答。

### 練習 1｜名字與年齡

讓使用者輸入名字與年齡，並輸出指定句型。

**輸入與輸出示例：**
```text
請輸入名字：Amy
請輸入年齡：16
Hello Amy, you are 16 years old.
```

**參考解答**

````python
print('Hello ', input('請輸入名字：'), ', you are ',
      input('請輸入年齡：'), ' years old.', sep='')
````

### 練習 2｜使用 `sep` 與 `end` 印出圖案

輸出以下圖案，**不可直接把整行 `* * *` 當成一個字串輸出**。

```text
* * *
* * *
* * *
```

**參考解答：使用 `sep` 與 `end`**

````python
print('*', '*', '*', sep=' ', end='\n')
print('*', '*', '*', sep=' ', end='\n')
print('*', '*', '*', sep=' ', end='\n')
````

### 練習 3｜連續輸出數字

使用**五個 `print()`**，輸出 `12345`，數字之間不能有空格或換行。

**參考解答**

````python
print(1, end='')
print(2, end='')
print(3, end='')
print(4, end='')
print(5, end='')
````

## 5. 本課重點整理

| 語法 | 功能 | 範例 |
|---|---|---|
| `print(x)` | 輸出內容 | `print('hi')` |
| `input()` | 讀取使用者輸入，回傳字串 | `input()` |
| `input('提示')` | 顯示提示並讀取輸入 | `input('名字：')` |
| `sep` | 設定多個輸出項目的分隔符號 | `print(1, 2, sep='-')` |
| `end` | 設定輸出結尾 | `print(1, end='')` |

**注意：** `input()` 回傳的是字串；日後若要進行數值計算，需要先轉換型別。
