# Lesson 02｜變數、資料型態與輸入輸出

## 學習目標

- 使用 `split()` 分割輸入，搭配多重變數賦值。
- 辨識 `int`、`str`、`float`、`bool` 與 `list`，並使用 `type()` 檢查型態。
- 使用 `int()` 將整數形式的字串轉換為整數。

> 本課使用 Python 3。遇到 `input()` 時，請在輸入框輸入範例資料並按 Enter。

## 1. 基本輸入與輸出

`input()` 讀取一行文字，回傳字串（`str`）；`print()` 可一次輸出多個值，預設以空格分隔。

**範例輸入：** `Apple`

**預期輸出：**
```text
Hi, Apple
Apple
```

````python
text = input()

print("Hi,", text)
print(text)
````

## 2. 分割輸入與多重賦值

### 2.1 使用 `split()`

不指定分隔符號時，`split()` 會以空白字元分割字串，並回傳串列（`list`）。

**範例輸入：** `10 20`

**預期輸出：** `['10', '20']`

````python
print(input().split())
````

### 2.2 將兩個輸入值分別存入變數

`a, b = input().split()` 將兩個字串分別存入 `a` 和 `b`。注意：此時它們是 `'10'` 與 `'20'`，而非整數 `10` 與 `20`。

**範例輸入：** `10 20`

**預期輸出：**
```text
10
20
```

這種寫法要求分割後恰好有兩個值。

````python
a, b = input().split()

print(a)
print(b)
````

## 3. 資料型態

### 3.1 數值加法、字串串接與串列

- 整數使用 `+` 進行加法。
- 字串使用 `+` 進行串接。
- 串列可以包含不同資料型態，也可以包含其他串列。

````python
print(123 + 123)
print("apple" + "banana")

items = ["a", 123, True, False, 12.30, [1, 23]]
print(items)
````

### 3.2 使用 `type()` 檢查資料型態

| 型態 | 意義 | 範例 |
|---|---|---|
| `int` | 整數 | `10` |
| `str` | 字串 | `'5.1'` |
| `float` | 浮點數 | `5.1` |
| `bool` | 布林值 | `True` |
| `list` | 串列 | `[10, '5.1', True]` |

即使字串內容看起來像數字，只要以引號包住，仍是字串。

````python
x = 10
y = "5.1"
z = True
items = [x, y, z]

print(type(x))
print(type(y))
print(type(z))
print(type(items))
````

## 4. 變數命名

變數名稱可以包含英文字母、數字及底線，但不能以數字開頭，也不能使用 Python 保留字。

`print_1` 是合法的變數名稱，與內建函式 `print()` 不同。

````python
print_1 = 1

print(print_1)
````

### Q3｜讀取兩個值

讓使用者輸入兩個數字，例如 `7 9`，分別存入 `a`、`b`，並輸出 `a` 和 `b`。

**範例輸入：** `7 9`

**預期輸出：** `7 9`

````python
# 參考解答
a, b = input().split()
print(a, b)
````

### Q4｜型態轉換與加法

承上題，將 `a` 和 `b` 轉成整數後，輸出 `a + b`。

**範例輸入：** `7 9`

**預期輸出：** `16`

提示：使用 `int()` 將字串轉換成整數。

````python
# 補充參考解答
a, b = input().split()
a = int(a)
b = int(b)
print(a + b)
````

### Q5｜挑戰題：輸入與文字輸出

**範例輸入：** `milk_tea 60`

**預期輸出：** `milk_tea 的價格是 60 元`

````python
# 補充參考解答
name, price = input().split()
print(name, "的價格是", price, "元")
````

## 6. 本課重點整理

| 語法 | 用途 |
|---|---|
| `input()` | 讀取一行字串 |
| `print()` | 輸出資料 |
| `input().split()` | 將一行輸入分割成字串串列 |
| `a, b = ...` | 將兩個值分別賦予兩個變數 |
| `type(x)` | 檢查資料型態 |
| `int(x)` | 將適當的值轉換為整數 |

**常見錯誤：** `'7' + '9'` 會得到字串 `'79'`，而 `int('7') + int('9')` 才會得到整數 `16`。
